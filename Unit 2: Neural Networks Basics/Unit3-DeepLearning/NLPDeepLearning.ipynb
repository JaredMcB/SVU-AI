```python
import pickle
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import confusion_matrix
import numpy as np
import pandas as pd

from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA

import nltk
import re

```

```python
# Get my review data
dataCH = pickle.load(open("../../FinalProject/DataCh.pkl", "rb"))
dataOB = pickle.load(open("../../FinalProject/DataOb.pkl", "rb"))
```

```python
del dataOB["Rating"]
del dataCH["Rating"]

dataCH["Source"] = 1
dataOB["Source"] = 0

data = pd.concat([dataCH, dataOB], ignore_index=True)
data
```

```python
# update vectorizer to remove English stop words and rebuild features
vectorizer = CountVectorizer(ngram_range=(1, 2), stop_words='english')
X_2gram = vectorizer.fit_transform(data["Review"])
```

```python
X_2gram.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_2gram, data["Source"], test_size=0.3, random_state=42)
```

```python
ml = MLPClassifier(hidden_layer_sizes=(128,64), max_iter=500, random_state=42)
ml.fit(X_train, y_train)
y_pred = ml.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
accuracy = np.trace(cm) / np.sum(cm)
print(f"Accuracy: {accuracy:.2f}")
```

```python
pca = PCA(n_components=50)
X_train_pca = pca.fit_transform(X_train.toarray())
X_test_pca = pca.transform(X_test.toarray())

mlp = MLPClassifier(hidden_layer_sizes=(128,64), max_iter=500, random_state=42)
mlp.fit(X_train_pca, y_train)
y_pred = mlp.predict(X_test_pca)
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
accuracy = np.trace(cm) / np.sum(cm)
print(f"Accuracy: {accuracy:.2f}")
```
